```python

import pandas as pd

data = {
    'Name': ['Arun', 'Priya', 'Kumar', 'Divya'],
    'Age': [20, 21, 22, 20],
    'Marks': [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print("\nFirst two rows:")
print(df.head(2))

print("\nDataFrame Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nStudent Names:")
print(df['Name'])

print("\nStudents with marks above 80:")
print(df[df['Marks'] > 80])

df['Result'] = ['Pass', 'Pass', 'Pass', 'Pass']
print("\nDataFrame after adding Result:")
print(df)
```
